You can expand any continuous function as a polynomials

$ f(x)=\sum_{n=0}^\infty \frac{1}{n!}f^{(n)}(x_0)\,(x-x_0)^n $

Here, $ f^{(n)}$ is the nth derivative and $ x_0 $ is the argument around which we expand the function
